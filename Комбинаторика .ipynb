### Задача 1.8
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

* Рассмотрим событие $A$ – призы **различны**

$P(A) = n^k = 10^5$

```python
from itertools import product

setItem = '0123456789'
a = set(product(setItem, repeat=5))
print('Всего комбинаций: {:d}.'.format(len(a)))
for item in a:
    print(item)
```

* Рассмотрим событие $В$ – призы **одинаковы**

$P(A) = {C_{10+5-1}^5} = {C_{14}^5} = \frac{14*13*12*11*10}{1*2*3*4*5}$

```python
from itertools import combinations_with_replacement

setItem = '0123456789'
k = 5

a = list(combinations_with_replacement(setItem, r=k))
print('Всего комбинаций: {:d}.'.format(len(a)))
for item in a:
    print(item)
```

### Задача 1.9
> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?

* Рассмотрим кол-во случаев $A$: существование семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза

$P_7(3;2;2) = \frac{7!}{3! \cdot 2! \cdot 2!}$

```python
import scipy.special as sc

n=7
n1=3
n2=2
n3=2
a = sc.factorial(n)/(sc.factorial(n1)*sc.factorial(n2)*sc.factorial(n3))
print(a)
```

### Задача 1.10
> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

* Рассмотрим событие $A$ - получение слова **ТОР**

$P(A) = \frac{m}{n} = \frac{1}{A_6^3} = \frac{1}{6*5*4}$

```python
import scipy.special as sc

n=6
k=3
round(1/sc.perm(n,k),4)
```

* Рассмотрим событие $B$ - получение слова **ТЕОРИЯ**

$P(B) = \frac{m}{n} = \frac{1}{P_6} = \frac{1}{6!}$

```python
import scipy.special as sc
n = 6
round(1 / sc.factorial(n), 4)
```

### Задача 1.11
> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

* Рассмотрим событие $B$ - получение слова **АНАНАС**

$P(B) = \frac{m}{n} = \frac{P_3*P_2}{P_6} = \frac{3! * 2!}{6!}$

```python
import scipy.special as sc

n1=3
n2=2
n3=6
a = sc.factorial(n1)*sc.factorial(n2)/sc.factorial(n3)
round(a,4)
```

### Задача 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

* Рассмотрим событие $A$ - никакие два лица одного
пола **не сядут рядом**.

$P(A) = \frac{m_1}{n} = \frac{2(5!)^2}{10!}$

```python
import scipy.special as sc

n=5
N=10
round(2*(sc.factorial(n)**2)/sc.factorial(N),5)
```

* Рассмотрим событие $B$ - супруги, образующие пять супружеских
пар, **сядут рядом**.

$P(B) = \frac{m_2}{n} = \frac{5!*2^5}{10!}$

```python
import scipy.special as sc

n=5
N=10
round(sc.factorial(n)*2**5/sc.factorial(N),5)
```

### Задача 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

* Рассмотрим событие $A$ - пассажиры попали в **два** купе.

$P(A) = \frac{m_1}{n} = \frac{C_9^2*C_8^7}{C_{36}^7}$

```python
import scipy.special as sc

round(sc.comb(9, 2) * sc.comb(8, 7) / sc.comb(36, 7), 7)
```

* Рассмотрим событие $B$ - пассажиры попали в **семь** купе.

$P(B) = \frac{m_2}{n} = \frac{C_9^7*4^7}{C_{36}^7}$

```python
import scipy.special as sc

round(sc.comb(9, 7) * 4**7 / sc.comb(36, 7), 5)
```

* Рассмотрим событие $D$ - пассажиры попали в **три** купе.

$P(D) = \frac{m_3}{n} = \frac{C_9^3*(6*C_4^4*C_4^2*C_4^1+3*C_4^3*C_4^3*C_4^1+3*C_4^3*C_4^2*C_4^2)}{C_{36}^7}$

```python
import scipy.special as sc

m3 = sc.comb(9, 3)*(6*sc.comb(4, 4)*sc.comb(4, 2)*sc.comb(4, 1)+ 3*sc.comb(4, 3)*sc.comb(4, 3)*sc.comb(4, 1)+3*sc.comb(4, 3)*sc.comb(4, 2)*sc.comb(4, 2))
round( m3 / sc.comb(36, 7), 5)
```
