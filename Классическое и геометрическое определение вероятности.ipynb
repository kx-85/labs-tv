### Задача 1.38
> Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева направо в порядке нумерации томов (от 1 до 5)?

* Рассмотрим вероятность $A$ – книги стоят слева направо в порядке нумерации томов (от 1 до 5).

$P(A) =\frac{n}{m} = \frac{n}{P_5} = \frac{1}{5!}$

```python
import scipy.special as sc

round(1 / sc.factorial(5), 4)
```

### Задача 1.43
> Наудачу взятый телефонный номер состоит из 5 цифр. Какова вероятность того, что в нем все цифры: а) различные; б) одинаковые; в) нечетные? Известно, что номер телефона не начинается с цифры ноль.

* Рассмотрим вероятность $A$ – все цифры **раличны**

$P(A) =\frac{n}{m} = \frac{9*9*8*7*6}{10^4*9}$

```python

9*9*8*7*6/(9*10**4)
```

* Рассмотрим вероятность $B$ – все цифры **одинаковые**
(11111,22222,33333,44444,55555,66666,77777,88888,99999)

$P(B) =\frac{n}{m} = \frac{9}{10^4*9}$

```python
9/(9*10**4)
```

* Рассмотрим вероятность $D$ – все цифры **нечетные**

$P(D) =\frac{n}{m} = \frac{5^5}{10^4*9}$

```python
round (5**5/(9*10**4),4)
```

### Задача 1.48
> В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения ll очков; б) выигрыша.

* Рассмотрим вероятность $A$ – выпадения **ll** очков.



```python

listItem = '123456'
numb = [int(i) for i in listItem]
f=0
for j in range(6):
    for k in range(6):
        for h in range(6):
            if numb[j]+numb[k]+numb[h]==11:
                f+=1
print(f)
round(f/6**3,4)
```

* Рассмотрим вероятность $B$ –выигрыша.

```python
#выигрыш >=11
listItem = '123456'
numb = [int(i) for i in listItem]
f=0
for j in range(6):
    for k in range(6):
        for h in range(6):
            if numb[j]+numb[k]+numb[h]>=11:
                f+=1
print(f)
round(f/6**3,4)
```
